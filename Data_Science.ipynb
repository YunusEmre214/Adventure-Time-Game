```python
#Veri Bilimi
import pandas as pd
```

```python
#1.Veri setini yükle
df=pd.read_csv("e_ticaret_veri_seti.csv")
```

```python
#2.Veri setinin ilk ve son satırları inncele
print(df.head())
```

```python
print(df.tail())
```

```python
#3.Veri setinin boyutunu ve sütunlarını kontrol etmek
row,column=df.shape
print(df.shape)
print(f"Row:{row}")
print(f"Column:{column}")
```

```python
#sütun isimleri
for i, column in enumerate(df.columns,start=1):
  print(f"{i}:{column}")
```

```python
#4.Veri setinin genel bilgisini inceleme
df.info()
```

```python
#5.Özet istatislikleri görüntüleme
df.describe()
```

```python
#6.Veri tiplerini kontrol etme
df.dtypes
```

```python
#7.Sayısal ve kategorik sütunların ayrılması
numeric_column=df.select_dtypes(include=["int64","float64"]).columns
categorical_column=df.select_dtypes(include=["object"]).columns
print(f"Numeric Column:{numeric_column}")
print(f"Categorical Column:{categorical_column}")
```

```python
#8.Benzersiz(unique) değer sayıların incelenmesi
for column in df.columns:
  print(f"{column}:{df[column].nunique()}")
```

```python
#Missing Value
#9.Eksik verileri tespit etme
missing_data=df.isnull().sum()
print(missing_data)
```

```python
#10.Eksik veri oranları hesaplama
missing_data_rate=(missing_data/len(df))*100
print(missing_data_rate)
```

```python
#11.Eksik veri içeren satırları incele
missing_data_row=df[df.isnull().any(axis=1)]
missing_data_row.head(10)
```
